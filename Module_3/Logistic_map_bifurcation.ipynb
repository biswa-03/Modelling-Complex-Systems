```python
def logistic_map(x, r):
    return r * x * (1 - x)
```

```python
# Time Series 
x0 = 0.2
r = 3.9
N = 100

x = np.empty(N+1)
x[0] = x0
for i in range(N):
    x[i+1] = logistic_map(x[i], r)
```

```python
plt.figure(figsize=(10,6)) 
plt.plot(x, marker='o', color='steelblue', linewidth=1.5)
plt.scatter(range(len(x)), x, s=20, color='tomato', zorder=3)
plt.title(r'Logistic Map ($X_{N+1} = rX_N(1 - X_N)$) Time Series (r={r})', fontsize=13)
plt.xlabel(" Iteration ")
plt.ylabel(" X_N ")
plt.tight_layout()
```

```python
# bifurcation Diagram
r_values = np.linspace(2.5, 4.0, 1000) # Generates 1000 r values from 2.5 to 4.0
transient = 1000 # Number of transient iterations to discard
samples = 100 # Number of samples to plot after transient

bif_r, bif_x = [], []
for r in r_values:
    x = 0.5 # Initial condition
    for i in range(transient + samples):
        x = logistic_map(x, r)
        if i >= transient: # Only plot after transient
            bif_r.append(r)
            bif_x.append(x)
```

```python
 # Bifurcation diagram
plt.figure(figsize=(10,6)) 
plt.scatter(bif_r, bif_x, s=0.3, color='steelblue', alpha=0.4)
plt.axvline(r, color='tomato', linewidth=1.0, linestyle='--', label=f"r = {r}")
plt.title("Bifurcation Diagram", fontsize=13)
plt.xlabel("r")
plt.ylabel("$X^*$")
plt.xlim(2.5, 4.0)
plt.ylim(0, 1)
plt.legend(fontsize=9)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
