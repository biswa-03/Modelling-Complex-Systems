```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def tent_map(x):
    return 1 - 2 * abs(x - 0.5)

# Parameters
x0 = 0.1
n  = 100

# Generate time series
x = np.empty(n + 1)
x[0] = x0
for i in range(n):
    x[i + 1] = tent_map(x[i])
```

```python
# Plot
plt.figure(figsize=(10, 4))
plt.plot(x, color='steelblue', linewidth=1.2)
plt.scatter(range(len(x)), x, s=12, color='tomato', zorder=3)
plt.title(r"Tent Map Time Series:  $X_{n+1} = 1 - 2|X_n - \frac{1}{2}|$", fontsize=13)
plt.xlabel("n (iteration)")
plt.ylabel("$X_n$")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
print("Done.")
```
