```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def shift_map(x):
    return (2 * x) % 1

# initialize parameters
x0 = 0.678247
n = 100

# generate time series
x = np.empty(n+1)
x[0] = x0
for i in range(n):
    x[i+1] = shift_map(x[i])
```

```python
plt.figure(figsize=(8,6))
plt.plot(x, color='steelblue', linewidth=1.2)
plt.scatter(range(len(x)), x, s=12, color='tomato', zorder=3)
plt.title(r"Shift Map Time Series:  $X_{n+1} = 2X_n$", fontsize=13)
plt.xlabel("n (iteration)")
plt.ylabel("$X_n$")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
print("Done.")
```
