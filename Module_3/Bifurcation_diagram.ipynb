#### Logistic Map

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def logistic_map(x, r):
    return r * x * (1 - x)
```

```python
#Time Series
x0 = 0.2
r  = 3.9
n  = 100

x = np.empty(n + 1)
x[0] = x0
for i in range(n):
    x[i + 1] = logistic_map(x[i], r)
```

```python
# Bifurcation Diagram
r_values  = np.linspace(2.5, 4.0, 1000)
transient = 500
samples   = 200

bif_r, bif_x = [], []
for r_val in r_values:
    xi = x0
    for _ in range(transient):
        xi = logistic_map(xi, r_val)
    for _ in range(samples):
        xi = logistic_map(xi, r_val)
        bif_r.append(r_val)
        bif_x.append(xi)
```

```python
# Plotting
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

# Time series
ax1.plot(x, color='steelblue', linewidth=1.2)
ax1.scatter(range(len(x)), x, s=12, color='tomato', zorder=3)
ax1.set_title(r"Logistic Map Time Series:  $X_{n+1} = rX_n(1 - X_n)$" + f"   (r = {r})", fontsize=13)
ax1.set_xlabel("n (iteration)")
ax1.set_ylabel("$X_n$")
ax1.set_ylim(-0.05, 1.05)
ax1.grid(True, linestyle='--', alpha=0.5)

# Bifurcation diagram
ax2.scatter(bif_r, bif_x, s=0.3, color='steelblue', alpha=0.4)
ax2.axvline(r, color='tomato', linewidth=1.0, linestyle='--', label=f"r = {r}")
ax2.set_title("Bifurcation Diagram", fontsize=13)
ax2.set_xlabel("r")
ax2.set_ylabel("$X^*$")
ax2.set_xlim(2.5, 4.0)
ax2.set_ylim(0, 1)
ax2.legend(fontsize=9)
ax2.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
print("Done.")
```

#### Sine Map
