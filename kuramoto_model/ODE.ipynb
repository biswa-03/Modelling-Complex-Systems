## Simulating Ordinary Ddifferential Equations

#### Simulating dx/dt = -x

```python
 # Importing modules and setting up variables

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy.integrate import solve_ivp

dt = 0.01   # Time step
T = 5   # Total time
steps = int(T/dt)
X = 5   # Initial condition

time = []
X_values = []
```

```python
t = 0
for i in range(steps):
    time.append(t)
    X_values.append(X)

    df = -X
    X = X + df*dt
    t = t + dt    
```

```python
plt.figure()
plt.plot(time, X_values)
plt.xlabel("Time")
plt.ylabel("X(t)")
plt.title("Plotting dX/dt = -X")
plt.show
```

#### Simulating a pendulum

```python
# Constants
g = 9.81      # gravity (m/s^2)
L = 1.0       # length of pendulum (m)
```

```python
# System of ODEs
def pendulum(t, y):
    theta, omega = y
    dtheta_dt = omega
    domega_dt = -(g/L) * np.sin(theta)
    return [dtheta_dt, domega_dt]

# Initial conditions
theta0 = np.pi / 3      # initial angle (60 degrees)
omega0 = 0.0            # initial angular velocity
y0 = [theta0, omega0]

# Time span
t_span = (0, 10)
t_eval = np.linspace(0, 10, 1000)

# Solve ODE
sol = solve_ivp(pendulum, t_span, y0, t_eval=t_eval)

# Plot results
plt.figure()
plt.plot(sol.t, sol.y[0])
plt.xlabel("Time (s)")
plt.ylabel("Angle (rad)")
plt.title("Nonlinear Pendulum Simulation")
plt.show()
```
