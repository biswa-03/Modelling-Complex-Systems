```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def logistic_map(x, r):
    return r * x * (1 - x)
```

```python
# Time Series 
x0 = 0.2
r = 3.9
N = 100

x = np.empty(N+1)
x[0] = x0
for i in range(N):
    x[i+1] = logistic_map(x[i], r)
```

```python
plt.figure(figsize=(10,6)) 
plt.plot(x, marker='o', color='steelblue', linewidth=1.5)
plt.scatter(range(len(x)), x, s=20, color='tomato', zorder=3)
plt.title(r'Logistic Map ($X_{N+1} = rX_N(1 - X_N)$) Time Series (r={r})', fontsize=13)
plt.xlabel(" Iteration ")
plt.ylabel(" X_N ")
plt.tight_layout()
```
