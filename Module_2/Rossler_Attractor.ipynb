## Rossler Attractor

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

### For a= 0.2, b=0.2, c=5.7

```python
# Parameters

a = 0.2
b = 0.2
c = 5.7
```

```python
# Rössler system
def rossler(t, state):
    
    x, y, z = state
    dxdt = -y - z
    dydt = x + a * y
    dzdt = b + z * (x - c)
    return [dxdt, dydt, dzdt]

# Time span
t_start = 0
t_end = 200
t_points = np.linspace(t_start, t_end, 50000)
```

```python

# Initial condition
initial_state = [1, 1, 1]

# Solve ODE
solution = solve_ivp(rossler, 
                     [t_start, t_end], 
                     initial_state, 
                     t_eval=t_points)

x = solution.y[0]
y = solution.y[1]
z = solution.y[2]

# 3D Plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')
ax.plot(x, y, z, lw=0.5)

ax.set_title("Rössler Attractor")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
# Time settings
t_start = 0
t_end = 150
t_points = np.linspace(t_start, t_end, 60000)
initial_state = [1, 1, 1]

# Different parameter sets (a, b, c)
parameter_sets = [
    (0.1, 0.1, 4.0),
    (0.1, 0.1, 6.0),
    (0.1, 0.1, 8.5),
    (0.1, 0.1, 12),
    (0.1, 0.1, 12.6),
    (0.1, 0.1, 15.4)
]

# Create figure
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(projection='3d', facecolor="k")

# Generate distinct colors automatically
colors = plt.cm.Set1(np.linspace(0, 1, len(parameter_sets)))

# Loop over parameter sets
for i, (a, b, c) in enumerate(parameter_sets):

    def rossler(t, state):
        x, y, z = state
        dxdt = -y - z
        dydt = x + a * y
        dzdt = b + z * (x - c)
        return [dxdt, dydt, dzdt]

    sol = solve_ivp(
        rossler,
        [t_start, t_end],
        initial_state,
        t_eval=t_points
    )

    x, y, z = sol.y
    ax.plot(x, y, z, color=colors[i], lw=0.7,
            label=f"a={a}, b={b}, c={c}")

# Labels

ax.set_title("Multiple Rössler Attractors")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.legend()

plt.show()
```
